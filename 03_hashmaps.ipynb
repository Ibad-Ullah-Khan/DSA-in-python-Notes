## __Hashing__
Hashing is a technique used to convert a key into a number so that data can be stored and retrieved quickly.
### __Why Hasing__
- 1: Fast Searching: Hashing lets us store and find data quickly without searching the entire collection.

| Hashing                               | HashMap                                 |
| ------------------------------------- | --------------------------------------- |
| A **technique/process**               | A **data structure**                    |
| Converts a key into a hash value      | Stores **key-value pairs**              |
| Uses a **hash function**              | Uses hashing internally                 |
| Helps decide where data should go     | Lets us store and retrieve data quickly |
| Example: `"Sara"` → `some hash value` | `{"Sara": 95}`                          |


```python
# Simplest example of using hashing

students = {
    "Ali": 90,
    "Ahmed": 85,
    "Sara": 95
}

students["Sara"]
```

### __Hash Function__

A hash function is a function that takes a key as input and converts it into a hash value (usually an integer).

### __Why do we need it__
- The hash value helps a HashMap decide where to store or find the data.

hash functions can be different.

## __Collision__

A collision happens when two different keys produce the same hash value.

```python
def hash_function(key):
    return key % 10
"""
12 % 10 = 2
22 % 10 = 2

12 ──→ 2
22 ──→ 2
      ↑
   Same location
"""
```

### __Techniques to handle collisions__

#### __1: Chaining__
Chaining is a collision-handling technique where multiple keys that have the same hash value are stored together at the same hash table location, usually in a linked list or another collection.

#### __Technique__
- Techiniques to reduce chain in chaining

##### __i) Rehashing__
Rehashing means creating a bigger hash table and inserting all existing elements into the new table using the new table size.

__Why__? Because when a hash table becomes too full, collisions increase, making searching slower.

##### __ii) 2. Convert the chain into a tree__

Instead of storing collided elements in a linked list, we can store them in a Binary Search Tree (BST):


```python
# Binary tree
"""
              32
           /     \
         22 \     52
        /    25  /  \
      12        42   62
"""
```

### __2: Open Addressing__

Open addressing is a collision-handling technique where, if the calculated position is already occupied, we find another empty position in the same hash table.

### __Techniques__
- Techinques to find another slot

#### __i) Linear probing__
Linear probing means:

When a collision occurs, check the next slot, then the next, until an empty slot is found.

The formula is:

 -             index=(h(k)+i)mod m

#### __ii) Quadratic Probing__
Quadratic probing handles a collision by checking positions with increasing quadratic jumps.

The formula is:

-            index=(h(k)+i^2) mod m
