# __Array__

An array is a linear data structure used to store multiple elements in a continous memory location.

### __Disadvantages__
- Fixed Size
- Homogenious(lack of flaxiblity)

```python
arr = [10,47,92,0]
arr[0]
```

### __Types of Array__
- Based on what the array stores
### __1: Primitive Array__

A primitive array stores the actual values directly in the array's memory.

The array contains the actual values.

```python
arr = [10, 20, 30, 40]
"""
Memory:
┌────┬────┬────┬────┐
│ 10 │ 20 │ 30 │ 40 │
└────┴────┴────┴────┘
"""
```

### __2: Referential Array__

A referential array stores references (addresses) to objects rather than storing the objects themselves.

List is a referential array

__Draw Backs__

- Extra memory 
- Speed(Slow)

```python
arr = [10, "Hello", 3.14, True]
""" 
arr
 ↓
┌──────────┬──────────┬──────────┬──────────┐
│ reference│ reference│ reference│ reference│
└────┬─────┴────┬─────┴────┬─────┴────┬─────┘
     ↓          ↓          ↓          ↓
    10       "Hello"      3.14      True
    
"""
```

### __Types of Array__
- Based on size
### __1: Static Array__
Size is fixed or difficult to change


```python
# Python doesn't have a true static array, but you can simulate one:

arr = [None] * 5

arr[0] = 10
arr[1] = 20
arr[2] = 30
arr[3] = 40
arr[4] = 50

print(arr)
```

### __2: Dynamic Array__
A Dynamic Array is an array whose size can automatically increase or decrease while the program is running.
- A dynamic array is an implementation technique in which a larger static array is allocated when the current one becomes full. The elements are copied to the new array, making the array appear to grow dynamically.

- Python List is a Dynamic Array

```python
arr = []

arr.append(10)
arr.append(20)
arr.append(30)
arr.append(40)
arr.append(50)
arr.append(60)

print(arr)
```
