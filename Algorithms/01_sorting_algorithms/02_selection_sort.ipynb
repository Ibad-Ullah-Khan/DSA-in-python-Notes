### __Selection Sort__

Selection Sort repeatedly finds the smallest element from the unsorted part and puts it at the beginning of that part.

### __Algorithm__


```python
""" 
1. **Start.**
2. Set `i = 0` as the starting position of the unsorted part.
3. Assume the element at position `i` is the **minimum element**.
4. Compare it with every element in the remaining unsorted part.
5. If a smaller element is found, update the **minimum position**.
6. After checking all remaining elements, swap the minimum element with the element at position `i`.
7. Move `i` to the next position.
8. Repeat steps 3–7 until the entire array is sorted.
9. **Stop.**
"""
```

### __Pseudocode__

```python
""" 
SelectionSort(A)
n = length(A)
for i = 0 → n - 2
    minIndex = i
    for j = i + 1 → n - 1
        if A[j] < A[minIndex]
            minIndex = j
    swap A[i], A[minIndex]
return A
"""
```

### __Time Complexity__

```python
""" 
O(n²)
"""
```

### __Space Complexity__

```python
""" 
O(1)
"""
```

### __Code__

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n-1):
        smallest = i
        for j in range(i+1, n):
            if arr[j] < arr[smallest]:
                smallest = j

        arr[i] , arr[smallest] = arr[smallest], arr[i]

    return arr

arr = [5,3,8,2,1]
print(selection_sort(arr))
```
