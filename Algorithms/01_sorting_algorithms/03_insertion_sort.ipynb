### __Insertion Sort__

Insertion Sort is a sorting algorithm that builds the sorted array one element at a time by taking each element and inserting it into its correct position in the already-sorted portion.

### __Algorithm__


```python
"""
1. Start.
2. Consider the **first element as sorted**.
3. Take the next element as the `key`.
4. Compare the `key` with elements in the sorted portion from right to left.
5. If an element is greater than the `key`, shift it one position to the right.
6. Continue until the correct position for the `key` is found.
7. Insert the `key` into that position.
8. Repeat steps 3–7 for all remaining elements.
9. The array is now sorted.
10. Stop.

"""
```

### __Pseudocode__

```python
""" 
InsertionSort(arr):
    n = length(arr)

    for i = 1 → n - 1:
        key = arr[i]
        j = i - 1

        while j >= 0 AND arr[j] > key:
            arr[j + 1] = arr[j]
            j = j - 1

        arr[j + 1] = key

    return arr
"""
```

### __Time Complexity__

```python
""" 
O(n²)
"""
```

### __Space Complexity__

```python
""" 
O(1)
"""
```

### __Code__

```python
def insertion_sort(arr):
    n = len(arr)

    for i in range(1, n):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr


nums = [5, 3, 8, 2, 1]

print(insertion_sort(nums))
```
