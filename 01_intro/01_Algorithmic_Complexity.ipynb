# __Algorithm Complexity:__

Algorithm Complexity is a measure of how efficient an algorithm is.

It tells us:

- How much time an algorithm takes to run.
- How much memory an algorithm uses.

So, algorithm complexity has two types:

* 1: Time Complexity
* 2: Space Complexity
### __Efficiency:__

How fast does our program run, and how much memory does it use?

# __1: Time Complexity:__

Time Complexity tells us how the number of operations grows as the input size (n) grows.
- It does not tell us the actual time in seconds.
- It tells us how the work (operations) increases when the input becomes larger.


### __Technique to measure Time Efficiency:__

### __1: Measuring Time:__
It means we run a program and measure how long it takes to finish.



```python
import time

start = time.time()
for i in range(1,101):
    print(i, end=' ')
print(f"\nTotal Time consume: {start - time.time()}")\

# But it is not a good approch to measure time complexity

"""
Disadvantages:
 
Time varies if implementation changes 
Different machines different time 
Does not work for extremely small input 
Time varies for different inputs, but can't establish a relationship 
"""
"""
Problem Solved:
Different Time for different algorithm
"""
```

### __3: Order of Growth__

Order of growth describes how the number of operations increases as the input size n increases.

- Big O is the notation (symbol) we use to represent the order of growth.
- Order of Growth is the most powerful technique because it compares algorithms based on how their work grows as the input size increases, regardless of machine speed, implementation details, or small constant differences.

### __How to calculate order of growth__

```python
#################################### Problem 1 ####################################

L = [1,2,3,4,5]

sum = 0
for i in L:
    sum = sum + i
print(sum)

product = 1
for i in L:
    product = product * i
print(product)


#################################### Problem 2 ####################################

L = [1,2,3,4,5]

for i in L:
    for j in L:
        print('{},{}'.format(i,j))


#################################### Problem 4 ####################################
```

# __2: Space Complexity:__

Space Complexity measures:

- How much memory an algorithm needs as the input size (n) increases.

Think of it this way:

- Time Complexity → How much work/time does the algorithm need?
- Space Complexity → How much memory/space does the algorithm need?

__Important distinction__

- When calculating Space Complexity, we are usually interested in the extra/auxiliary memory used by the algorithm, excluding the input itself.

__Def:__ Space Complexity is the amount of extra memory an algorithm requires as the input size increases.
