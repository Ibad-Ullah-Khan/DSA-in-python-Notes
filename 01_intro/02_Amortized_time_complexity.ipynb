# Amortized Time Complexity

## 1. What is Amortized Time Complexity?

Amortized time complexity is the average cost per operation over a sequence of operations, even if some individual operations are expensive.

It helps us understand the total cost of performing many operations rather than looking at just one operation.

## 2. Example: Inserting Elements into a Dynamic Array

A **dynamic array** automatically increases its capacity when it becomes full.

Suppose:

* Initial capacity = 2
* When the array becomes full, its capacity doubles.
* Each insertion normally takes O(1) time.
* When resizing occurs, existing elements must be copied, which takes O(n) time.

### Step-by-step example

| Operation | Array                            | Cost | Explanation                |
| --------- | -------------------------------- | ---: | -------------------------- |
| Insert 10 | [10]                             |    1 | Insert directly            |
| Insert 20 | [10, 20]                         |    1 | Insert directly            |
| Insert 30 | [10, 20, 30]                     |    3 | Copy 2 elements + insert 1 |
| Insert 40 | [10, 20, 30, 40]                 |    1 | Insert directly            |
| Insert 50 | [10, 20, 30, 40, 50]             |    5 | Copy 4 elements + insert 1 |
| Insert 60 | [10, 20, 30, 40, 50, 60]         |    1 | Insert directly            |
| Insert 70 | [10, 20, 30, 40, 50, 60, 70]     |    1 | Insert directly            |
| Insert 80 | [10, 20, 30, 40, 50, 60, 70, 80] |    1 | Insert directly            |

*Note: The cost is measured in simplified units of work, not exact CPU instructions.*

## 3. Calculating the Average Cost

### Step 1: Calculate the total cost

Add the costs of all 8 insertions:

$$
1+1+3+1+5+1+1+1=14
$$

Total cost = 14

### Step 2: Count the operations

We inserted 8 elements.

Number of operations = 8

### Step 3: Calculate the average cost

$$
\text{Average Cost}=\frac{\text{Total Cost}}{\text{Number of Operations}}
$$

$$
=\frac{14}{8}=1.75
$$

This means that, in this example, each insertion took an average of 1.75 units of work.

It does not mean every insertion takes 1.75 units. Some insertions take 1 unit, while resizing insertions take more.

## 4. Why is Amortized Time O(1)?

In a dynamic array that doubles its capacity:

* Most insertions take O(1).
* Occasionally, an insertion takes O(n) because elements must be copied.
* However, resizing does not happen on every insertion.

Over a long sequence of n insertions, the total cost is O(n).

Therefore:

$$
\text{Amortized Cost}=\frac{O(n)}{n}=O(1)
$$

So, **appending an element to a dynamic array has O(1) amortized time complexity**, even though an individual append operation can take O(n).

## 5. Difference Between Worst-Case and Amortized Time

| Worst-case time                                                 | Amortized time                                                 |
| --------------------------------------------------------------- | -------------------------------------------------------------- |
| Measures the cost of one expensive operation in the worst case. | Measures the cost per operation over a sequence of operations. |
| Dynamic-array append can be O(n).                               | Dynamic-array append is O(1) amortized.                        |

## 6. Key Takeaway

Amortized time complexity helps us understand why an operation can be efficient overall, even when some individual operations are expensive.

**Remember:**

* Worst-case time for one dynamic-array append: O(n)
* Amortized time for dynamic-array append: O(1)
* Total cost of n appends: O(n)
